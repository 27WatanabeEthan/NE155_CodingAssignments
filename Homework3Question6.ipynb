```python
import numpy as np
import math
```

#### #1
$f(x) = \sqrt{x}$ at $x=0.5$

```python
# forward difference
print("Forward Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = np.sqrt(0.5 + h) - np.sqrt(0.5)
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# backward difference
print("Backward Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = np.sqrt(0.5) - np.sqrt(0.5 - h)
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# centered difference
print("Centered Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = np.sqrt(0.5 + h) - np.sqrt(0.5 - h)
    solution /= 2*h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# complex step approximation
print("Complex Step Approximation")
for i in range(5):
    h = 2**(-i-1)
    solution = np.imag(np.sqrt(0.5 + 1j*h))
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

#### #2
$f(x) = \arctan{(x^2 - 0.9x + 2)}$ at $x=0.5$

```python
# forward difference
print("Forward Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = np.arctan((0.5+h)**2 - 0.9*(0.5+h) + 2) - np.arctan(0.5**2 - 0.9*0.5 + 2)
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# backward difference
print("Backward Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = np.arctan(0.5**2 - 0.9*0.5 + 2) - np.arctan((0.5-h)**2 - 0.9*(0.5-h) + 2)
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# centered difference
print("Centered Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = np.arctan((0.5+h)**2 - 0.9*(0.5+h) + 2) - np.arctan((0.5-h)**2 - 0.9*(0.5-h) + 2)
    solution /= 2*h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# complex step approximation
print("Complex Step Approximation")
for i in range(5):
    h = 2**(-i-1)
    solution = np.imag(np.arctan((0.5+1j*h)**2 - 0.9*(0.5+1j*h) + 2))
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

#### #3
$J_0(x) = \sum_{m=0}^{\infty} \frac{(-1)^m}{m! \Gamma(m+1)} \left( \frac{x}{2} \right)^{2m}$ at $x=1$

```python
def J_0(x, terms=50): # 50 terms is good enough
    sol = 0
    for m in range(terms):
        sol += (-1)**m * (x/2)**(2*m) / (math.factorial(m))**2
    return sol
```

```python
# forward difference
print("Forward Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = J_0(1+h) - J_0(1)
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# backward difference
print("Backward Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = J_0(1) - J_0(1-h)
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# centered difference
print("Centered Difference")
for i in range(5):
    h = 2**(-i-1)
    solution = J_0(1+h) - J_0(1-h)
    solution /= 2*h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```

```python
# complex step approximation
print("Complex Step Approximation")
for i in range(5):
    h = 2**(-i-1)
    solution = np.imag(J_0(1 + 1j*h))
    solution /= h
    print(f"h = 2^({-i-1}): f'(x) = {solution}")
```
