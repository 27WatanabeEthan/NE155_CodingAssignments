```python
import numpy as np
import matplotlib.pyplot as plt
from matrix_solvers import gaussian_elimination
```

```python
xs = np.arange(0, 1.21, 0.01)
ys0 = np.cos(xs)

ys1 = -0.400435*xs**2 - 0.050846*xs + 1

ys2 = 0.0922412*xs**3 - 0.565112*xs**2 + 0.0139388*xs + 1

plt.plot(xs, ys0, color="#4582FF", label="cos(x)")
plt.plot(xs, ys1, color="green", ls="--", label="2nd Degree")
plt.plot(xs, ys2, color="red", label="3rd Degree")

plt.title("Lagrange Interpolation")
plt.grid(True, which="both", ls="-", alpha=0.3)
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# Absolute Errors
rel_error1 = np.abs(ys1 - ys0) / ys0
rel_error2 = np.abs(ys2 - ys0) / ys0

plt.plot(xs, rel_error1, color="green", ls="--", label="2nd Degree")
plt.plot(xs, rel_error2, color="red", label="3rd Degree")

plt.title("Relative Interpolation Error")
plt.yscale('log')
plt.grid(True, which="both", ls="-", alpha=0.3)
plt.xlabel("x")
plt.ylabel("Relative Error")
plt.legend()
plt.show()
```
