```python
import matrix_solvers as ms
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([
    [4, -1,  0,  0,  0],
    [-1, 4, -1,  0,  0],
    [0, -1,  4, -1,  0],
    [0,  0, -1,  4, -1],
    [0,  0,  0, -1,  4]
])
b = np.array([100, 100, 100, 100, 100])
guess = np.zeros(5)
```

```python
# 0 < omega < 2
xs = np.arange(0.1, 2, 0.1)
ys = np.array([])
for x in xs:
    iterations, solutions = ms.gauss_seidel(A, b, guess, sor=x)
    ys = np.append(ys, iterations)
# print(ys)
plt.scatter(xs, ys, color="#5F00A8")
plt.xlabel("ω")
plt.ylabel("Iterations")
plt.title("ω vs. # of iterations")
plt.show()
```
