```python
from matrix_solvers import gaussian_elimination
import numpy as np
```

```python
chi = np.array([0.57500, 0.42500, 0.00000, 0.00000])
nuSf = np.array([0.00480, 0.00060, 0.00855, 0.09255])
Sigma_a = np.array([0.00490, 0.00280, 0.03050, 0.12100])
Sigma_s = np.array([0.08310, 0.05850, 0.06510, 0.00000])
A = np.zeros([4, 4])
for i in range(4):
    A[i, i] = Sigma_a[i] + Sigma_s[i] - chi[i]*nuSf[i]
    for j in range(4):
        if i == j:
            pass
        else:
            A[i, j] = -chi[i]*nuSf[j]
    if i > 0:
        A[i, i-1] -= Sigma_s[i-1]

Q = np.array([1e12, 0, 0, 0])

# print(A)
sol, matrix = gaussian_elimination(A, Q)
print(sol)
# print(matrix)
```
