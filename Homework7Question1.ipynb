```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
def LCG(N, seed=0):
    rand_nums = np.empty(N)
    for i in range(N):
        if i == 0:
            rand_nums[i] = (5*seed + 3) % 16
        else:
            rand_nums[i] = (5*rand_nums[i-1] + 3) % 16
    return rand_nums
```

```python
print(LCG(10, seed=7))
```
